```python
#  Boundary Conditions: Password Length Validator

while True:
  password = input("Enter password:")

#  This is to see the conditions of validation
  if 8 <= len(password) <= 15:
      print("The password is valid.")
      break
# This is to tell users limitations of validation
  else:
    print("The password is too long or short. Please try again.")
```

```python
# Multiple Decision Paths: Loan Eligibiility Checker

while True:
  credit_score = int(input("Enter Credit Score: "))
  annual_income = int(input("Enter Annual Income: "))
  yearsatcurrentjob = int (input("Enter Years at Current Job: "))

# This is to see the conditions of eligibility
  if credit_score >= 700 and annual_income >= 30000 and yearsatcurrentjob >= 2:
    print("Loan Approved")
    break

# This is to tell the users limitations of eligibility
  else:
    print("Loan Denied")
    break
```
