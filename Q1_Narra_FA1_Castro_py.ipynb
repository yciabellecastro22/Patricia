```python
# This is the predetermined library
import math

# This is where to input values for each variable
x1 = float(input("Enter x1: "))
x2 = float(input("Enter x2: "))
y1 = float(input("Enter y1: "))
y2 = float(input("Enter y2: "))

# This distance formula will calculate the distance between the two points when the values are entered
distance = math.sqrt(math.pow(x2 - x1, 2) + math.pow(y2 - y1, 2))

# This is where the rounded distance shows on the calculator
print("The distance between the two points is:", round (distance,2))
```
